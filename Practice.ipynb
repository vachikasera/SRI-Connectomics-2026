```python
def average(numbers):
    if not numbers:
        return None
    return sum(numbers) / len(numbers)

print(average([1, 2, 3, 4, 5]))
```

I found three taste-bristle sensory neurons: LEGNP_T2.3453 (SNch09), LEGNP_T2.3593 (SNch05), and LEGNP_T2.3677 (SNch05). These neurons are classified as chemosensory sensory neurons that detect taste information from the mesothoracic leg and transmit signals to downstream neurons in the nervous system. The neurons used different neurotransmitters, including acetylcholine and glutamate, suggesting they may communicate taste information through different signaling pathways. Biologically, these connections likely help the fly process chemical cues in its environment, allowing it to identify food sources and avoid potentially harmful substances.
